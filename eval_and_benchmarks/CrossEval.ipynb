# Cross-corpus evaluation — Vienna4x22 & Batik-plays-Mozart

Mazurka-trained checkpoints (5 folds) tested on two unseen corpora as **test-only** splits. Pack the test HDF5s with:

```bash
python pytorch/data_preprocess.py --mode pack_h5_vienna --sample_rate 22050
python pytorch/data_preprocess.py --mode pack_h5_batik  --sample_rate 22050
```

See **[CrossEval_README.md](CrossEval_README.md)** for the prior-distribution table, why Mozart_K331 is excluded, and the coarse 3-class metric.

**Reported metrics** (per fold):
- `dynamic_f1`         — fine-grained 5-class (pp/p/mf/f/ff)
- `dynamic_coarse_f1`  — 3-class (quiet=pp+p / medium=mf / loud=f+ff)  ← cross-corpus comparable
- `change_point_f1`, `beat_f1`, `downbeat_f1`

## 0. Setup — run once at top

````python
import os, subprocess

# Project root (where pytorch/ lives). This notebook lives one level below it
# in eval_and_benchmarks/, so the parent dir is the project root regardless
# of where Jupyter was launched from.
PROJ = os.path.abspath(os.path.join(os.getcwd(),
       "" if os.path.isdir("pytorch") else ".."))
assert os.path.isdir(os.path.join(PROJ, "pytorch")), f"pytorch/ not found under {PROJ}"

# Workspace that holds the packed test HDF5s and where outputs/ will land.
TEST_WS    = "/media/datadisk/home/22828187/zhanh/202509_bark_data/workspaces"
OUT        = f"{TEST_WS}/outputs"
VIENNA_CSV = f"{TEST_WS}/split_csvs/vienna_sr22050_test.csv"
BATIK_CSV  = f"{TEST_WS}/split_csvs/batik_sr22050_test.csv"

# Checkpoint roots - 5-fold ckpts for MultiTask and SingleCNN live
# in separate directories on this machine.
CKPT_MULTI  = "/media/datadisk/home/22828187/zhanh/202509_icassp_data/workspaces/mazurka_ckpts/ckpt_multi"
CKPT_SINGLE = "/media/datadisk/home/22828187/zhanh/202509_icassp_data/workspaces/mazurka_ckpts/ckpt_single"

INFER = os.path.join(PROJ, "pytorch", "inference.py")
EVAL  = os.path.join(PROJ, "pytorch", "final_evaluation.py")

def run_test(ckpt_pattern: str, dataset: str, csv_path: str,
             folds=range(5), min_epoch=8):
    """5-fold mazurka-trained model -> test on the SAME Vienna/Batik split.
    `ckpt_pattern` must contain `{f}` for the fold index.
    """
    assert dataset in {"vienna", "batik"}
    for f in folds:
        common = [
            "--ckpt_dir", ckpt_pattern.format(f=f),
            "--infer_split_csv", csv_path,
            "--output_dir", OUT,
            "--valf1_rank", "1",
            "--min_epoch", str(min_epoch),
            "--overrides", f"dataset.name={dataset},exp.workspace={TEST_WS}",
        ]
        # cwd=PROJ so the scripts find their hydra config + pytorch/ siblings
        subprocess.run(["python", INFER, *common], check=True, cwd=PROJ)
        subprocess.run(["python", EVAL,  *common], check=True, cwd=PROJ)
````

# 1. MultiTaskCNN — BSSL (sone)

````python
# ===== Vienna4x22 — MultiTaskCNN + BSSL =====
CKPT = f"{CKPT_MULTI}/formal latent8 mmoe8 5x5-sone-MultiTaskCNN-['dynamic', 'change_point', 'beat', 'downbeat']-fold{{f}}"
run_test(CKPT, dataset="vienna", csv_path=VIENNA_CSV)
````

````python
# ===== Batik-plays-Mozart — MultiTaskCNN + BSSL =====
CKPT = f"{CKPT_MULTI}/formal latent8 mmoe8 5x5-sone-MultiTaskCNN-['dynamic', 'change_point', 'beat', 'downbeat']-fold{{f}}"
run_test(CKPT, dataset="batik", csv_path=BATIK_CSV)
````
